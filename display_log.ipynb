```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python

df = pd.read_csv('log.csv')
fig, ax1 = plt.subplots()


# First line (left y-axis)
ax1.plot(df['Timeline'], df['Error_X'], 'b-', label='Line 1')
ax1.set_xlabel('time (s)')
ax1.set_ylabel('Error X (in)', color='b')
ax1.tick_params(axis='y', labelcolor='b')
ax1.axhline(0, color='blue', linestyle='--', linewidth=0.5)
max_abs = max(abs(ax1.get_ylim()[0]), abs(ax1.get_ylim()[1]))
ax1.set_ylim(-max_abs, max_abs)


# Second line (right y-axis)
ax2 = ax1.twinx()
ax2.plot(df['Timeline'], df['PID_X'], 'r-', label='Line 2')
ax2.set_ylabel('PID Output', color='r')
ax2.tick_params(axis='y', labelcolor='r')
ax2.axhline(-15, color='red', linestyle='--', linewidth=0.5)
ax2.axhline(15, color='red', linestyle='--', linewidth=0.5)
max_abs = max(abs(ax2.get_ylim()[0]), abs(ax2.get_ylim()[1]))
max_abs = min(max_abs, 128)
ax2.set_ylim(-max_abs, max_abs)

plt.show()
```
